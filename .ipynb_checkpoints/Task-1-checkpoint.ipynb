### Python String Functions

#### Example-1

```python
text_string = 'vigneshbabu'

print('Python Strings \n---------------')

print('Hi', text_string)
print("Welcome to 'Python' Learning")
print(''' I'm happy 
          to join with you 
          to learn AI''')

print('\nGet 0th index:', text_string[0])
print('\nGet all strings')
for x in 'John':
    print(x)

print('\nLength of string:', len(text_string))
print('\nExists check:', 'vig' in text_string)
print('\nNot Exists check:', 'babu' not in text_string)

print('\ncheck the string with if case')
if 'john' in text_string:
    print('"john" not exits in given string')
else:
    print("'john' exits in given string")


print('\nSlicing strings', '\n---------------')

# return v to h from text_string
print(text_string[0:3])

# slice to the start from v
print(text_string[:7])

# slice to the end from b
print(text_string[7:])

#slice with negative indexing
print(text_string[-4:])

print('\nString modification \n------------------------------')

print(text_string.lower())
print(text_string.upper())
print(text_string.capitalize())
print(text_string.strip())
print(text_string.replace('vignesh', 'ganesh'))

new_string = "Hello, World"
print(new_string.split(","))

print('\n String concatenation', '\n----------------')

string1 = 'Hello'
string2 = 'world'

res = string1 + ' ' + string2

print(res)
print(string1, string2)


print('\n String Format', '\n--------------')
age = 33
print(f"My name is {text_string}, I'm {age}.")
price = 100
qty = 3
print(f"Price is {price * qty:.2f} rupees")

print('\n Escape characters', '\n-------------------')

print(f"It\'s {type(age)} data type")
print('Hello\tWorld')
print('Hello\rWorld')
print('Hello \bWorld')
print('Hello\nWorld')
```

#### Example-2

```python
# string methods

user_input = " Python Programming "

print('\n Given input:', user_input)

print('\n title:', user_input.title())

print('\n swap:', user_input.swapcase())

print('\n strip:', user_input.strip())

print('\n lstrip:', user_input.lstrip())

print('\n rstrip:', user_input.rstrip())

print('\n count:', user_input.count('a'))

try:
    print('\n index:', user_input.lower().index("p"))
except ValueError:
    print('\n Substring not found')
# print('\n index:', user_input.index("p"))

print('\n find:', user_input.find('programm'))

print('\n alpha only:', user_input.isalpha())

print('\n digit only:', user_input.isdigit())

print('\n alpha numeric:', user_input.isalnum())

print('\n starts with python:', user_input.lower().startswith('python'))

print('\n end with ing:', user_input.lower().endswith('ing'))

print('\n Check space:', user_input.isspace())

print('\n zfill:', "100".zfill(5))

join_str = "-".join(["2026", "07", "12"])

print('\n join_str:', join_str)

print("apple, orange, banana".split(','))
```
