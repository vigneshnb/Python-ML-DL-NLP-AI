### Word guessing game

```python
secret_word = 'python'
guessed_word = ['_'] * len(secret_word)
attempts = len(secret_word)

while attempts > 0 and '_' in guessed_word:d

    print('\nWord: ', ' '.join(guessed_word))
    print('\nNo of attempts left: ', attempts)

    try:
        guess = input('\nEnter a guess character: ').lower()

        if len(guess) !=1 and not guess.isalpha():
            print('\nAllow only alphabets')

        if guess in guessed_word:
            print('\nGuessed word is duplicate')

        if guess in secret_word:

            for index in range(len(secret_word)):
                if secret_word[index] == guess:
                    guessed_word[index] = guess
                    print('\nCorrect Guess')

        else:
            attempts -=1
            print('\nWrong Guess')

    except Exception as error:
        print('Error', error)

if "_" not in guessed_word:
    print('Congratulations, you won')
else:
    print('Game over, a correct word is:', secret_word)
```
