```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Titanic.csv')

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='Sex')

plt.title('Passenger Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('No of Passengers')
plt.show()
```

```python
plt.figure(figsize=(7,7))

pclass_values = df['Pclass'].value_counts().sort_values()

plt.pie(
    pclass_values,
    labels=['1st class', '2nd class', '3rd class'],
    autopct="%1.1f%%",
    startangle=90
)

plt.title('Distribution by Passenger Class')
plt.show()
```

```python
plt.figure(figsize=(7,5))

embarked_counts = df['Embarked'].value_counts()

sns.barplot(
    x=embarked_counts.values,
    y=embarked_counts.index
)

plt.title('Distribution by Boarding Port')
plt.xlabel('No of passengers')
plt.ylabel('Boarding point')

plt.show()
```

```python
plt.figure(figsize=(7,7))

survival_counts = df['Survived'].value_counts()

plt.pie(
    survival_counts,
    labels=['Did not Survive', 'Survived'],
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops=dict(width=0.4)
)

plt.title('Distribution by survival')

plt.show()
```

```python
plt.figure(figsize=(10,7))

sns.histplot(
    data=df,
    x='Age',
    kde=True
)

plt.title('Distribution by Age')
plt.xlabel('Age')
plt.ylabel('No of Passengers')
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(data=df, x='Age')

plt.title('Distribution by Age and Outliers')
plt.show()
```

```python
plt.figure(figsize=(8,7))

sns.countplot(data=df, x='SibSp')

plt.show()
```

```python
plt.figure(figsize=(8,7))

sns.countplot(data=df, x='Parch')

plt.show()
```

```python
df = pd.read_csv('Titanic.csv')

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='Sex',
    hue='Survived'
)

plt.title('Survival based on Gender')
plt.xlabel('Gender')
plt.ylabel('No of Passengers')
plt.legend(
    ['No', 'Yes'],
    title='Survival'
)
plt.show()
```

```python
survival_rate = df.groupby('Pclass')['Survived'].mean().reset_index()

sns.barplot(
    data=survival_rate,
    x='Pclass',
    y='Survived'
)

plt.ylim(0, 1)
plt.show()
```

```python
sns.boxplot(
    data=df,
    x='Survived',
    y='Age'
)

plt.show()
```

```python
sns.countplot(
    data=df,
    x='Embarked',
    hue='Survived'
)
```

```python
sns.barplot(
    data=df,
    x='Pclass',
    y='Survived',
    hue='Sex'
)
```

```python
plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="Sex",
    y="Age",
    hue="Survived"
)

plt.title("Age Distribution by Gender and Survival")
plt.xlabel("Gender")
plt.ylabel("Age")

plt.legend(
    title="Survived",
    labels=["No", "Yes"]
)

plt.show()
```
