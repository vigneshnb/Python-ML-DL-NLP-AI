```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

titanic_data = pd.read_csv('Titanic.csv')
print('Rows & Columns', titanic_data.shape)
print()
print('Data Types')
print()
print(titanic_data.dtypes)
print()
print('See Null Data')
print()
print(titanic_data.isnull().sum())
print()
print(titanic_data.describe())
print()
print(titanic_data.duplicated().sum())
print()
```

### Univariate Analysis

```python
# Gender Distribution - Bar chart

plt.figure(figsize=(7 , 5))

sns.countplot(data=titanic_data, x = 'Sex')

plt.title('Passenger Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('No of Passengers')
plt.show()
```

```python
# Class Distribution - Pie chart

pclass_counts = titanic_data['Pclass'].value_counts().sort_index()

plt.figure(figsize=(7, 7))

plt.pie(
    x=pclass_counts,
    labels=['1st class', '2nd class', '3rd class'],
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Passenger Distribution by Class')
plt.show()
```

```python
# Boarding port Distribution - Horizontal Bar chart

plt.figure(figsize=(7, 5))
    
embarked_counts = titanic_data['Embarked'].value_counts()

sns.barplot(
    x=embarked_counts.values,
    y=embarked_counts.index
)

plt.title('Passengers by Boarding port')
plt.ylabel('Embarked')
plt.xlabel('No of passengers')

plt.show()
```

```python
# Distribution by Survival

survival_counts = titanic_data['Survived'].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    survival_counts,
    labels=['Did not Survive', 'Survived'],
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops=dict(width=0.4)
)

plt.title('Survival Distribution')
plt.show()
```

```python
# Age Distribution - Histplot

plt.figure(figsize=(9, 5))
sns.histplot(
    data=titanic_data,
    x='Age',
    bins=20,
    kde=True
)

plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('No of passengers')
plt.show()
```

```python
# Age Distribution and outliers - boxplot

plt.figure(figsize=(8, 4))

sns.boxplot(
    data=titanic_data,
    x='Age'
)

plt.title('Age distribution and outliers')
plt.xlabel('Age')

plt.show()
```

```python
# Passenger by No of sibling / spouses

plt.figure(figsize=(8, 5))

sns.countplot(
    data=titanic_data,
    x='SibSp'
)

plt.title('Passenger by No of Siblings / Spouses')
plt.xlabel('No of Siblings / Spouses')
plt.ylabel('No of Passengers')
plt.show()
```

```python
# Passenger by No of Parents / children

plt.figure(figsize=(8, 5))

sns.countplot(
    data=titanic_data,
    x='Parch'
)

plt.title('Passenger by No of Parents / children')
plt.xlabel('No of Parents / children')
plt.ylabel('No of Passengers')
plt.show()
```

### Bivariate Analysis

```python
# Gender vs Survival — Grouped Bar Chart
titanic_data = pd.read_csv('Titanic.csv')
plt.figure(figsize=(8, 5))

sns.countplot(
    data=titanic_data,
    x="Sex",
    hue="Survived"
)

plt.title("Survival Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.legend(
    title="Survived",
    labels=["No", "Yes"]
)

plt.show()
```

```python
# Passenger Class vs Survival Rate — Bar Chart

survival_rate = (
    titanic_data.groupby("Pclass")["Survived"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=survival_rate,
    x="Pclass",
    y="Survived"
)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")

plt.ylim(0, 1)

plt.show()
```

```python
# Age vs Survival — Box Plot

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=titanic_data,
    x="Survived",
    y="Age"
)

plt.title("Age Distribution by Survival Status")
plt.xlabel("Survival Status")
plt.ylabel("Age")

plt.xticks(
    [0, 1],
    ["Did Not Survive", "Survived"]
)

plt.show()
```

```python
# Embarked vs Survival — Grouped Bar Chart

plt.figure(figsize=(8, 5))

sns.countplot(
    data=titanic_data,
    x="Embarked",
    hue="Survived"
)

plt.title("Survival Distribution by Boarding Port")
plt.xlabel("Boarding Port")
plt.ylabel("Number of Passengers")

plt.legend(
    title="Survived",
    labels=["No", "Yes"]
)

plt.show()
```

### Multivariate Analysis

```python
# Gender + Passenger Class + Survival - Grouped Bar Chart

plt.figure(figsize=(9, 6))

sns.barplot(
    data=titanic_data,
    x="Pclass",
    y="Survived",
    hue="Sex"
)

plt.title("Survival Rate by Passenger Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")

plt.ylim(0, 1)

plt.gca().yaxis.set_major_formatter(
    plt.matplotlib.ticker.PercentFormatter(1)
)

plt.legend(title="Gender")

plt.show()
```

```python
# Age + Gender + Survival - Box Plot

plt.figure(figsize=(9, 6))

sns.boxplot(
    data=titanic_data,
    x="Sex",
    y="Age",
    hue="Survived"
)

plt.title("Age Distribution by Gender and Survival")
plt.xlabel("Gender")
plt.ylabel("Age")

plt.legend(
    title="Survived",
    labels=["No", "Yes"]
)

plt.show()
```

```python
# Age + Fare + Survival - Scatter Plot

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=titanic_data,
    x="Age",
    y="Fare",
    hue="Survived",
    size="Pclass",
    sizes=(40, 200),
    alpha=0.7
)

plt.title("Age, Fare and Survival by Passenger Class")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.legend(
    title="Survival / Class"
)

plt.show()
```

```python
# Passenger Class + Gender + Embarked + Survival - Facet Grid

g = sns.catplot(
    data=titanic_data,
    x="Pclass",
    y="Survived",
    col="Embarked",
    hue="Sex",
    kind="bar",
    height=5,
    aspect=0.9
)

g.set_axis_labels(
    "Passenger Class",
    "Survival Rate"
)

g.set_titles("Boarding Port: {col_name}")

g.set(ylim=(0, 1))

plt.show()
```

```python
# Correlation Heatmap

numeric_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare"
]

correlation = titanic_data[numeric_columns].corr()

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Between Numerical Variables")

plt.show()
```

```python
# Pair Plot — Overall Relationship

sns.pairplot(
    titanic_data[
        ["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare"]
    ],
    hue="Survived",
    diag_kind="hist"
)

plt.show()
```
