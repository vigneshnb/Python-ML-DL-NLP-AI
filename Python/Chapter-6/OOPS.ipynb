### Bank account management system

```python
# Class & Objects
class BankAccount():

    def __init__(self, account_number, customer_name, balance):
        self.account_number = account_number
        self.customer_name = customer_name
        self.balance = balance

    def deposit(self, balance):
        self.balance += balance

    def get_balance(self):
        return self.balance

    def display_account_details(self):
        print(f"Account number: {self.account_number}")
        print(f"Customer name: {self.customer_name}")
        print(f"Balance: {self.balance}")


bank_account = BankAccount('0000001', 'Vignesh', 50000)
bank_account.customer_name = 'Vignesh babu'
bank_account.balance = 50000

bank_account.deposit(500) 
bank_account.display_account_details()
bank_account.get_balance()
```

```python
# Inherietence

class SavingsAccount(BankAccount):

    def get_interest_calculation(self):
        return self.balance * 1.10

savings_account = SavingsAccount('000002', 'Babu', 10000)
savings_account.get_interest_calculation()
```

```python
# Encapsulation

class BankAccount1():
    def __init__(self, account_number, customer_name, balance):
        self.account_number = account_number
        self.customer_name = customer_name
        self.__balance = balance #Protected

    def get_balance(self):
        return self.__balance

    def deposit(self, balance):
        self.__balance += balance

account_det = BankAccount1('0001', 'Vignesh', 20000)
account_det.customer_name = 'Ganesh'
account_det.__balance = 50000 # Value won't update

account_det.get_balance()
```

```python
# Polymorphism

class SavingsAccount(BankAccount):

    def calculate_interest(self):
        return self.get_balance() * 0.05

class CurrentAccount(BankAccount):

    def calculate_interest(self):
        return self.get_balance() * 0.01


sa = SavingsAccount('0001', 'Vignesh', 5000)
ca = CurrentAccount('0002', 'Ganesh', 5000)

print(sa.calculate_interest())
print(ca.calculate_interest())
```

```python
# Abstraction

from abc import ABC, abstractmethod

class BankAccount2(ABC):

    def __init__(self, account_number, customer_name, balance):
        self.account_number = account_number
        self.customer_name = customer_name
        self.balance = balance


    @abstractmethod
    def calculate_interset(self):
        pass

class Savings(BankAccount2):

    def calculate_interset(self):
        return self.balance * 0.10
        
ba = Savings('0001', 'Vig', 5000)
ba.calculate_interset()
```

```python
# Apply all oops concepts in the same file.

from abc import ABC, abstractmethod

class BankAccounts(ABC):
    def __init__(self, account_number, customer_name, balance):
        self.account_number = account_number
        self.customer_name = customer_name
        self.__balance = balance

    #Encapsulation
    def deposit(self, amount):

        if amount > 0:
            self.__balance += amount
            print('Amount deposited successfully')
        else:
            print('Invalid amount to deposit')

    def widthdraw(self, amount):

        if amount < 0:
            print('Invalid amount')
        elif amount > self.__balance:
            print('Insufficient balance')
        else:
            self.__balance -= amount
            print('Amount widthdrawn successfully')

    def get_balance(self):
        return self.__balance

    def display_account(self): 
        print("\nAccount Details") 
        print("----------------------") 
        print("Account Number:", self.account_number) 
        print("Customer Name:", self.customer_name) 
        print("Balance:", self.__balance)

    #Abstraction
    @abstractmethod
    def calculate_interest(self):
        pass

#Inheritence
class SavingsAccounts(BankAccounts):
    #Polymorphism
    def calculate_interest(self):
        return self.get_balance() * 0.05

class CurrentAccounts(BankAccounts):
    #Polymorphism
    def calculate_interest(self):
        return self.get_balance() * 0.01

saving = SavingsAccounts('0001', 'Vignesh', 50000)
current = CurrentAccounts('0002', 'Ganesh', 50000)

saving.deposit(5000)
saving.widthdraw(2000)
saving.display_account()
print()
current.deposit(5000)
current.widthdraw(2000)
current.display_account()
print()
print('Savings Account Interst:', saving.calculate_interest())
print('Current Account Interst:', current.calculate_interest())
```
