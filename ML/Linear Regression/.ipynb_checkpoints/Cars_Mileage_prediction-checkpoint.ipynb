```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression

# Data Collection
cars_data = pd.read_csv('mtcars.csv')
cars_data

# Data Understanding
cars_data.shape #(32, 12)
cars_data.isna().sum()
cars_data.dtypes
cars_data.duplicated()
cars_data.describe()
cars_data.info()
cars_data
```

### Check Assumptions

1. 

```python
# Data Preparation

X = cars_data[['cyl','hp','wt','am','gear','carb']]
y = cars_data['mpg']
```

```python
# Model Building

linear_model = LinearRegression()
```

```python
# Model Training

linear_int = linear_model.fit(X, y)
linear_int
```

```python
# Model Testing

y_pred = linear_int.predict(X)
y_pred
```
