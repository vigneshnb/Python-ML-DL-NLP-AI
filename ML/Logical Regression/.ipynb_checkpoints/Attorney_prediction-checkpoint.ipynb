```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
import warnings
warnings.filterwarnings('ignore')

# Data Collection
climants_data = pd.read_csv('climants.csv')

# Data Understanding
climants_data.shape
climants_data.isna().sum()
climants_data.describe()
climants_data.dtypes

# Data Preparation
climants_data.dropna(inplace=True)

X = climants_data.drop(['CASENUM', 'ATTORNEY'], axis=1)
y = climants_data['ATTORNEY']

# Model Building
logistic_model = LogisticRegression()

# Model Training
logistic_model.fit(X, y)

# Model Testing
y_predict = logistic_model.predict(X)
y_predict

# Model Evaluation
from sklearn.metrics import accuracy_score

accuracy_score(y, y_predict)

# Model Deployment
from pickle import dump, load

dump(obj=logistic_model, file=open(file='logistic.pkl', mode='wb'))

logistic_intl = load(file=open('logistic.pkl', 'rb'))
logistic_intl.predict(X)
```

```python
# Train Split Test

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Model Building
logistic_model = LogisticRegression()
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.20, shuffle=True, random_state=10)

# Model Training
logistic_model.fit(x_train, y_train)

# Model Testing
y_predict_train = logistic_model.predict(x_train)
y_predict_test = logistic_model.predict(x_test)

# Model Evaluation
print(accuracy_score(y_train, y_predict_train))
print(accuracy_score(y_test, y_predict_test))
```
