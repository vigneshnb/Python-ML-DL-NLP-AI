```python
# Square

rows = 5

for i in range(rows):
    for j in range(rows):
        print('*', end=' ')
    print()
```

```python
# Right angled Triangle

rows=5

for i in range(1, rows, 1):
    for j in range(i):
        print('*', end=' ')
    print()
```

```python
# Inverted right-angled triangle

rows=5

for i in range(rows, 0, -1):
    for j in range(i):
        print('*', end=' ')
    print()
```

```python
# Pyramid pattern

rows=5

for i in range(rows):
    print(' ' * (rows-i-1), end=' ')
    print('*' * (2*i + 1))
```

```python
# Inverted Pyramid pattern

rows=5

for i in range(rows-1, -1, -1):
    print(' ' * (rows-i-1), end = ' ')
    print('*' * (2*i+1))
```

```python
# Diamond pattern

rows = 5

for i in range(rows):
    print(' ' * (rows-i-1), end=' ')
    print('*' * (2*i+1))

for i in range(rows-2, -1, -1):
    print(' ' * (rows-i-1), end=' ')
    print('*' * (2*i+1))
```

```python
# Right-aligned triangle
rows = 5

for i in range(1, rows + 1):
    print(' ' * (rows - i), end='')
    print('*' * i)
```

```python
# Hallow Square

rows = 5

for i in range(rows):
    for j in range(rows):
        if i==0 or i==rows-1 or j==0 or j==rows-1:
            print('*', end='')
        else:
            print(' ', end= '')
    print()
```

```python
rows = 5

for i in range(1, rows+1):
    for j in range(i):
        print('*', end='')

    for j in range(2 * (rows-i)):
        print(' ',  end='')

    for j in range(i):
        print('*', end='')
        
    print()

for i in range(rows -1, 0, -1):
    for j in range(i):
        print('*', end='')

    for j in range(2*(rows-i)):
        print(' ', end='')

    for j in range(i):
        print('*', end='')

    print()
```

```python
# Hourglass star pattern
n = 5

for i in range(n):
    for j in range(i):
        print(' ', end='')

    for j in range(2 * (n-i)-1):
        print('*', end='')

    print()

for i in range(n-2, -1, -1):
    for j in range(i):
        print(' ', end='')

    for j in range(2 * (n-i)-1):
        print('*', end='')

    print()
```

```python
# X-shaped star pattern
n = 5

for i in range(n):
    for j in range(n):
        if i==j or i+j == n -1:
            print('*', end='')
        else:
            print(' ', end='')
    print()
        
```

```python
# Rhombus star pattern
n=5

for i in range(n):
    for j in range(n-i-1):
        print(' ', end='')

    for j in range(n):
        print('*', end='')

    print()
        
```
