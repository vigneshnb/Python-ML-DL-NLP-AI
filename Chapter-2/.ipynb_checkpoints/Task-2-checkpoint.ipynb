### DataFrames using dictionaries

#### Example - 1

```python
# Student Data
import pandas as pd

student_dict = {
    'Id': ['0001', '0002', '0003', '0004', '0005'],
    'Name': ['Akash','Bala','Chandru','Dinesh','Eswar'],
    'Tamil': [90, 80, 70, 60, 50],
    'Maths': [99, 89, 79, 60, 55],
    'Science': [98, 60, 70, 75, 50],
    'Grade': ['A', 'B', 'A', 'C', 'E']
}

print('Student Data')

pd.DataFrame(student_dict)
```

#### Example - 2

```python
#Employee data

employee_dict = {
    'Name': ['Akash','Bala','Chandru','Dinesh','Eswar'],
    'Degree': ['BE', 'BSC', 'BCOM', 'MBA', 'ME'],
    'Role': ['Senior Developer', 'Tester', 'Tester', 'Business Analyst', 'Project Manager'],
    'Joining Date': ['01-01-2021', '01-01-2021', '01-01-2021', '01-01-2019', '01-01-2016'],
    'Experience': ['5yrs', '5yrs', '5yrs', '7yrs', '10yrs'],
    'Salary': [100000, 25000, 25000, 70000, 200000],
}

print('Employee Data')

pd.DataFrame(employee_dict)
```

```python
# Product data

product_dict = {
    'Product Id': ['A0001', 'B0001', 'C0001', 'D0001', 'E0001', 'F0001', 'G0001'],
    'Name': ['Artificial Neckalce', 'Bag', 'Cycle', 'Diamond ring', 'Electric kettle', 'Fan', 'Geyser'],
    'Amount': [500, 800, 12000, 50000, 3500, 4000, 11000],
    'Qty': [10, 5, 20, 15, 30, 25, 40],
    'Return Allowed': ['Yes', 'No', 'Yes', 'No', 'Yes', 'Yes', 'No']
}

print('Product Data')

pd.DataFrame(product_dict)
```
