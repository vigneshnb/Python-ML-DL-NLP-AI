### Python Data structures

#### Example - List

```python
# List items are ordered, changeable and allow duplicates

fruits = ['Apple', 'Organge', 'Grapes']

print('\n Length of list:', len(fruits))

print('\n DataType', type(fruits))

dataList = ['Apple', True, False, 0, 1, 0.50]
print('\n different data types:', dataList)

dataList = list(('Apple', True, False, 0, 1))
print('\n List constructor:', dataList)

# Access List 
if 'Apple' in fruits:
    print('\n Apple exists in list')

if 'banana' not in fruits:
    print('\n banana not in list')

print('\n Get second item:', fruits[1])

print('\n Get All items', fruits[:])

print('\n Get first 2 items:', fruits[0:2])

print('\n Get last 2 items:', fruits[-2:])

print('\n Get first 2 items use of negative index:', fruits[-3:-1])

print('\n Get last item alone:', fruits[-1])

# Change List
fruits[0] = 'Lemon'
print('\n Change first item alone:', fruits)

fruits[0:2] = ['Apple', 'Papaya']
print('\n Change first 2 items:', fruits)

fruits[3:] = ['Mango']
print('\n Add new one at last postion:', fruits)

fruits[0:2] = ['Watermelon']
print('\n Change first item and remove 2nd items:', fruits)

#Add items
fruits.append('Jackfruit')
print(fruits)

fruits.extend(['Gowa'])
print(fruits)

fruits.insert(0, 'pomogrenate')
print(fruits)

#Remove items
fruits.remove('pomogrenate')
print('\n Remove item based on item name:', fruits)

fruits.pop(0)
print('\n Remove first item:', fruits)

fruits.pop()
print('\n Remove last item:', fruits)

del fruits[0]
print('\n Delete first item:', fruits)

fruits.clear()
print('\n Clear all items:', fruits)

del fruits
print('\n Delete list')

fruits = ['pomogrenate', 'Watermelon', 'Grapes']

#Loop items
print('\n List items')
for x in fruits:
    print('', x)

print('\n List item indexes')
for x in range(len(fruits)):
    print('', x)

print('\n List items using while')
i = 0
while i < len(fruits):
    print('', fruits[i])
    i += 1

#List shortcut code
print('\n List items using short code')
[x for x in fruits]

print('\n List items except pomogrenate')
[x for x in fruits if x != 'pomogrenate']

fruits.append('Gowa')
print('\n List items replace pomogrenate into Mango')
[x if x != 'Gowa' else 'Mango' for x in fruits]

#sort items
fruits.sort()
print('\n Sort items:', fruits)

fruits.append('apple')
fruits.sort(key=str.lower)
print('\n Sort items with lower fun:', fruits)

fruits.sort(reverse=True)
print('\n Sort items by reverse fun:', fruits)

fruits.reverse()
print('\n Sort items by reverse method:', fruits)

#copy items
fruits1 = fruits.copy()
print('\n Copy items by copy method:', fruits1)

fruits2 = list(fruits)
print('\n Copy items by list constructor:', fruits2)

fruits3 = []
for x in fruits:
    fruits3.append(x)
print('\n Copy items by for loop append:', fruits3)

#join list
fruits1 = ['Apple']
fruits2 = ['Banana']
fruits3 = fruits1 + fruits2
print('\n Join items by concatenation', fruits3)

fruits1.extend(fruits2)
print('\n Join items by extend method', fruits1)

#methods
print('\n Count items', fruits1.count('Apple'))
print('\n Find index', fruits1.index('Apple'))
```

#### Example - Tuples

```python
# Ordered, Unchangeable, Allow duplicates
fruits = ('Apple', 'Orange', 'Banana')

print('\n Fruits:', fruits)

single_tuple = ('Apple',)

print('\n Single Item:', single_tuple)

print('\n Length of tuple:', fruits)

print('\n DataType:', type(fruits))

tuple_items = tuple(('Apple', 'Orange'))

print('\n Tuple constructor:', tuple_items)

#Access Tuples
print('\n Get first item:', fruits[0])

print('\n Get last item:', fruits[-1])

print('\n Get range items:', fruits[0:2])

print('\n Get items from negative range:', fruits[-3:-1])

if 'Apple' in fruits:
    print('\n Apple Exists in Tuples list')

#Update Tuples
# we won't update tuples directly, we can convert it to list and update
fruits_lst = list(fruits)
fruits_lst[0] = 'App'
fruits = tuple(fruits_lst)
print('\n Update first item:', fruits)
print(fruits_lst)

fruits_lst.remove('App')
fruits = tuple(fruits_lst)
print('\n Remove 1st item:', fruits)

fruits_lst.append('Watermelon')
fruits = tuple(fruits_lst)
print('\n Added item in tuples:', fruits)

fruits_lst.insert(0, 'Apple')
fruits = tuple(fruits_lst)
print('\n Added item in first position:', fruits)

del fruits
print('\n Delete tuples')

fruits = ('Apple', 'Orange', 'Banana')

#Unpack tuples

x, y, z = fruits

print('\n Unpacked items:', x, y, z)

x, *y = fruits

print('\n Unpacked items use of Asterisk*:', x, y)

#Loop Tuples

print('\n Iterate tuples using for')
for x in fruits:
    print(' ', x)
    
print('\n Iterate tuples using while')
i = 0
while i < len(fruits):
    print(' ', fruits[i])

    i+=1

#Join Tuples
tuple1 = ('Apple',)
tuple2 = ('Orange',)
tuple3 = tuple1 + tuple2
print('\n join 2 tuples', tuple3)
print('\n multiply tuples:', tuple1*2)

#Tuple methods
print('\n Index of item', fruits.index('Apple'))
print('\n Count of item', fruits.count('Apple'))
```

#### Example - Set

```python
# Unordered, Unchanged, Not allow duplicates
fruits = {'apple', 'orange', 'banana'}

print('\n Length of set:', len(fruits))

print('\n Data type:', type(fruits))

print('\n Constructor set:', set((True, 0, 'Apple', False)))

#Access set items
if 'apple' in fruits:
    print('\n apple exists in fruits')

# Add items
fruits.add('Watermelon')
print('\n Add new item', fruits)

fruits.update({'Papaya'})
print('\n Add new item use of update', fruits)

fruits.update(['Mango'])
print('\n Add new item use of update list', fruits)

# Remove items
fruits.remove('Watermelon')
print('\n Remove Watermelon from set', fruits)

fruits.pop()
print('\n Remove first item from set', fruits)

fruits.discard('apple')
print('\n Remove apple from set using discard', fruits)

fruits.clear()
print('\n Clear all items', fruits)

del fruits
print('\n Delete set')

fruits = {'apple', 'orange', 'banana'}

#loop set
print('\n Iterate fruits')
for x in fruits:
    print(' ', x)

#join sets
fruits2 = {'watermelon', 'apple', 'mango'}
fruits3 = {'grapes', 'pomogrenate', 'watermelon'}
fruits4 = ['chekoo', 'watermelon']

# union & update - It's return items from all sets, and exclude duplicates
fruits.update(fruits2)
print('\n Join set using update', fruits)

print('\n Join set using union', fruits.union(fruits2, fruits3, fruits4))

print('\n Join set using |', fruits2 | fruits3)

#intersection - It keeps only the duplicates
print('\n Join set using intersection', fruits.intersection(fruits2))
print('\n Join set using &', fruits & fruits2)

fruits.intersection_update(fruits3)
print('\n Join with update set using intersection_update', fruits)

#difference
# It will return the items from the 1st set, that are not present in other set

fruits = {'apple', 'orange', 'banana'}
fruits2 = {'watermelon', 'apple', 'mango'}
fruits3 = {'grapes', 'pomogrenate', 'watermelon'}
fruits4 = ['chekoo', 'watermelon']

print('\n Join with difference', fruits.difference(fruits2, fruits3))
print('\n Join with difference', fruits - fruits2)
fruits.difference_update(fruits2, fruits3)
print('\n Join with difference', fruits)

fruits = {'apple', 'orange', 'banana'}

#symmetric_difference - It will keep only the elements, that are not present in both sets
print('\n Join with symmetric_difference', fruits.symmetric_difference(fruits2))
```

#### Example - Dictionary

```python
# Ordered, Changeable, Not allow duplicates

student_dict = {
    'name': 'vignesh',
    'degree':  'BE',
    'department': 'CSE',
    'cgpa': 70.35
}

print('', student_dict)

print('\n Get the student name:', student_dict['name'])

print('\n Length of dict:', len(student_dict))

print('\n Find Data type:', type(student_dict))

my_dict = dict({'Country': 'India'})
print('\n Constructor', my_dict)

#Access items
if 'name' in student_dict:
    print('\n Key (name) exists in dict')

print('\n Get cgpa:', student_dict['cgpa'])

print('\n Get degree using get:', student_dict.get('degree'))

print('\n Get keys:', student_dict.keys())

print('\n Get values:', student_dict.values())

print('\n Get items:', student_dict.items())

#Update items
student_dict['year'] = '2009-2013'

print('\n Update new item:', student_dict)

student_dict.update({'pass_out_year': 2013})

print('\n Update new item using update:', student_dict)

#Add items
student_dict.update({'arrears' : 'NIL'})
print('\n Add new item using update:', student_dict)

#Remove items
student_dict.pop('arrears')
print('\n Remove arrear item:', student_dict)

student_dict.popitem()
print('\n Remove last item:', student_dict)

del student_dict['degree']
print('\n Remove degree item using del:', student_dict)

student_dict.clear()
print('\n Clear dict:', student_dict)

del student_dict
print('\n Delete dict')

student_dict = {
    'name': 'vignesh',
    'degree':  'BE',
    'department': 'CSE',
    'cgpa': 70.35
}

#Loop dictionaries

print('\n Get keys using for')
for x in student_dict:
    print('', x)

print('\n Get values using for')
for x in student_dict.values():
    print('', x)

print('\n Get keys & values using for')
for x, y in student_dict.items():
    print('', x, y)

#Copy dictionaries
my_dict = student_dict.copy()
print('\n Copy dict using copy', my_dict)

mydict = dict(student_dict)
print('\n Copy dict using consructor', my_dict)

#Nested dict

mystudents = {
  "student1" : {
    "name" : "Emil",
    "year" : 2004
  },
  "student2" : {
    "name" : "Tobias",
    "year" : 2007
  },
  "student3" : {
    "name" : "Linus",
    "year" : 2011
  }
}

print('\n Iterate nested dict')
for x, obj in mystudents.items():
    print('\n', x)
    for y in obj:
        print(' ' + y + ':', obj[y])
```
