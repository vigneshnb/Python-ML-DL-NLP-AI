#### 1.	Write a program to take a name and age from the user and print a greeting message.

#### 2.	Create a list of numbers and print only even numbers using a for loop.

```python
even_number_list = []

for i in range(1, 11):
    if i % 2 == 0:
        even_number_list.append(i)

print(even_number_list)
```

#### 3.	Write a program that handles division by zero using try-except.

```python
try:
    divident = int(input('Enter Divident number'))
    divisor = int(input('Enter Divisor number'))
    result = divident/divisor
except ZeroDivisionError:
    print('Given number not divisible by zero')
except ValueError:
    print('Allow only positive numbers')
except:
    print('Something went wrong')

print('The resule of quotient is:', result)
```

#### 4.	Given a string, count how many vowels it has.

```python
input_string = input('Enter word to find vowels count')
count = 0

for x in input_string.lower():
    if x in 'aeiou':
        count += 1

print(f'Have {count} vowels in given word')
        
```

```python
number_list = [1,2,3]

reverse_list = []

for i in range(len(number_list) -1, -1, -1):
    reverse_list.append(number_list[i])

print(reverse_list)

# i = len(number_list) - 1
# while i >= 0:
#     reverse_list.append(number_list[i])
#     i -= 1

# print(reverse_list)
```

#### 9.	Replace all missing values in a DataFrame with the column’s mean.

```python
employee_data = pd.read_csv(r'E:\jupyter\Chapter-4\employee_data.csv')

df = pd.DataFrame(employee_data) 
print("Original DataFrame:\n")
print(df)

age_mean = employee_data['Age'].mean().round(2)
salary_mean = employee_data['Salary'].mean().round(2)

df.fillna({'Age': age_mean, 'Salary': salary_mean}, inplace=True)

print("\nDataFrame After Replacing Missing Values:\n")
print(df)
```

#### 10.	Filter rows in a DataFrame where the "Age" column is greater than 30.

```python
emp_data = pd.read_csv('employee_data.csv')

df = pd.DataFrame(employee_data) 
print("Original DataFrame:\n")
print(df)

df = df[df['Age'] > 30]
print("\nDataFrame After Filtering:\n")
print(df)

pd.DataFrame(df)
```
