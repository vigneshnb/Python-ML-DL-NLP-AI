```python
# Merge

data1 = {'key': ['K0', 'K1', 'K2', 'K3'],
         'Name':['Jai', 'Princi', 'Gaurav', 'Anuj'],
        'Age':[27, 24, 22, 32],}

data2 = {'key': ['K0', 'K1', 'K2', 'K3'],
         'Address':['Nagpur', 'Kanpur', 'Allahabad', 'Kannuaj'],
        'Qualification':['Btech', 'B.A', 'Bcom', 'B.hons']}

df = pd.DataFrame(data1, index=[0,1,2,3])
df2 = pd.DataFrame(data2, index=[4,5,6,7])

df3 = pd.merge(df, df2, on='key')
print(df3)
print()

data1 = {'key': ['K0', 'K1', 'K2', 'K3'],
         'key1': ['K0', 'K1', 'K0', 'K1'],
         'Name':['Jai', 'Princi', 'Gaurav', 'Anuj'],
        'Age':[27, 24, 22, 32],}

data2 = {'key': ['K0', 'K1', 'K2', 'K3'],
         'key1': ['K0', 'K0', 'K0', 'K0'],
         'Address':['Nagpur', 'Kanpur', 'Allahabad', 'Kannuaj'],
        'Qualification':['Btech', 'B.A', 'Bcom', 'B.hons']}

df = pd.DataFrame(data1, index=[0,1,2,3])
df2 = pd.DataFrame(data2, index=[4,5,6,7])

df3 = pd.merge(df, df2, on=['key','key1'])
print(df3)
print()

df4 = pd.merge(df, df2, how='inner', on=['key','key1'])
print(df4)
print('inner')
print()

df4 = pd.merge(df, df2, how='outer', on=['key','key1'])
print(df4)
print('outer')
print()

df5 = pd.merge(df, df2, how='left', on=['key','key1'])
print(df5)
print('left')
print()

df6 = pd.merge(df, df2, how='right', on=['key','key1'])
print(df6)
print('right')
print()
```
