### Rock, Paper, Scissors Game

```python
import random

while True:
    try:
        choices = ['rock','paper','scissors']

        computer_choice = random.choice(choices)

        user_choice = input('\nEnter Rock, Paper or Scissors').lower()

        print('\nYou choose:', user_choice)
        print('\nComputer choose:', computer_choice)

        if user_choice not in choices:
            print('\nInvalid Choice')
        elif user_choice == computer_choice:
            print('It\'s Tie')
        elif (user_choice == 'rock' and computer_choice == 'scissors' or \
               user_choice == 'paper' and computer_choice == 'rock' or \
               user_choice == 'scissors' and computer_choice == 'paper'):
            print('\nYou Win\'s')
        else:
            print('\nComputer Win\'s')

        play_again = input('\nDo you wants to play again? yes/no').lower()


        if play_again != 'yes':
            print('\nThank you, Come back to play again')
            break

        
    except Exception as error:
        print('Something went wrong', str(error))
```
