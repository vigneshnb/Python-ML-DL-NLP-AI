### while loop

#### Example-1

```python
# Print numbers from 1 to 5

i = 1

while i<= 5:
    print(i)
    i+=1
```

```python
# print sum of first five natural numbers

i = 1
total = 0

while i<=5:
    total += i
    i+=1

print(total)
```

### if else

#### Example-1

```python
# Check given number is even or odd

number = int(input('Enter the number'))

if number % 2 == 0:
    print('Given number is even')
else:
    print('Given number is Odd')
```

```python
# Check the vote eligibilty

age = int(input('Enter your age'))

if age >= 18:
    print('You\'re eligible to vote')
else:
    print('You\'re not eligible to vote')
```

### if elif else

#### Example-1

```python
# Student grade system

mark = int(input('Enter the mark with in 100'))

if mark >= 90:
    print('Grade A')
elif mark >= 75:
    print('Grade B')
elif mark >= 50:
    print('Grade C')
else:
    print('Fail')
```

#### Example-2

```python
# Traffic singnal pattern

signal_clr = input('Enter the color (Red, Yellow, Green')

if signal_clr.lower() == 'red':
    print('Stop')
elif signal_clr.lower() == 'yellow':
    print('Get ready')
elif signal_clr.lower() == 'green':
    print('Let\'s Go')
else:
    print('Invalid input')
```

### User defined function with for loop and if condition

#### Example-1

```python
# Get even numbers

def get_even_numbers(n):
    print('Even numbers\n')
    for i in range(1, n+1):
        if i % 2 == 0:
            print(i)
```

```python
get_even_numbers(20)
```

#### Example-2

```python
# Display student who are pass

def check_pass():

    students = [
        ('Vignesh', 90),
        ('Ganesh', 75),
        ('Vignesh', 50),
        ('Srinivas', 40),
    ]

    for name, marks in students:

        if marks >=50:
            print(name, marks)
        else:
            print(name, 'Fail')
```

```python
check_pass()
```

### break statement

#### Example-1

```python
# stop printing number at 6

for i in range(1,11):
    if i == 6:
        print('Stop at 6')
        break
    else:
        print(i)
```

#### Example-2

```python
# Stop searching when product is found

products = ['Desktop','Laptop','Mac','Mobile']

for x in products:

    if x == 'Mac':
        print('Product is found:', x)
        break
    else:
        print('Searching Product....', "'Mac'")
```

### continue statement

#### Example-1

```python
# Skip printing number as 5

for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```

```python
#### Example-2

# Skip the Student 'Vignesh'

students = ['arun', 'bala', 'chandru', 'vignesh', 'vinay']

for x in students:
    if x == 'vignesh':
        continue

    print(x)
```

### for loop with if statement

#### Example-1

```python
# Print Odd numbers

print('Odd numbers\n')
for i in range(1,10):

    if i % 2 > 0:
        print(i)
```

#### Example-2

```python
# Get fruits starts with M

fruits = ['Apple', 'Mango', 'Melon', 'Mangustan', 'Grapes']

for x in fruits:
    if x.startswith('M'):
        print(x)
```

### for loop with not in operator

#### Example-1

```python
# Print available students

students = ['arun','bala','vicky']
absent_students = ['vicky']

for x in students:
    if x not in absent_students:
        print(x)
```

```python
# Print available products

products = ['laptop', 'mouse', 'keyboard']
sold_out = ['keyboard']

for x in products:
    if x not in sold_out:
        print(x)
```

### if with in operator

### if with not in operator

#### Example-1

```python
# Check whether fruit not exists

fruits = ['Apple', 'Mango', 'Melon', 'Mangustan', 'Grapes']

if 'Lemon' not in fruits:
    print('Lemon not Exists')
```

#### Example-2

```python
# Check whether character not exists in given word

word = 'python'

if 'r' not in word:
    print('r not exists in given word')
```
