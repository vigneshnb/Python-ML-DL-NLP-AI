```python
import pandas as pd
import keras
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from keras.layers import Dense
from keras.models import Sequential
```

```python
di_data = pd.read_csv(r'E:\jupyter\DL\Chapter-9 (External Dataset ANN Classification)\Docs\diabetes.csv')
di_data

# di_data = pd.DataFrame(diabetic_data)
# di_data
```

```python
X = di_data.drop('Outcome', axis=1)
y = di_data['Outcome']

X.shape, y.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, shuffle=True, stratify=y, test_size=0.2)
```

```python
scalar = StandardScaler()
X_train = scalar.fit_transform(X_train)
X_test = scalar.transform(X_test)
```

```python
model = Sequential()
model.add(Dense(12, input_dim=8, activation='relu', kernel_initializer='uniform'))
model.add(Dense(8, activation='relu', kernel_initializer='uniform'))
model.add(Dense(1, activation='sigmoid', kernel_initializer='uniform'))

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model.fit(X_train, y_train, epochs=50, batch_size=16, validation_split=0.2)
```

```python
y_pred = model.predict(X_test)
y_pred = (y_pred > 0.60).astype(int)
y_pred
```

```python
model.save('Diabetes_intelligence.h5')
```
