```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import keras
```

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()
```

```python
X_train.shape, y_train.shape, X_test.shape, y_test.shape
```

```python
pd.set_option('display.max_columns', 30)
pd.DataFrame(X_train[10000])
```

```python
plt.imshow(X_train[10000])
```

```python
X_train = X_train/255.0
X_test = X_test/255.0
```

```python
X_train
```

```python
X_train.min(), X_train.max()
```

```python
X_test.min(), X_test.max()
```

```python
from keras.models import Sequential
from keras.layers import Flatten, Dense
```

```python
model = Sequential()
model.add(layer=Flatten())
model.add(Dense(units=100, activation='relu'))
model.add(Dense(10, activation='softmax'))

model.compile(optimizer='rmsprop', loss='sparse_categorical_crossentropy')
```

```python
model.fit(x=X_train,
    y=y_train,
    batch_size=32,
    epochs=10,
    validation_data=(X_test, y_test))
```

```python
model.summary()
```

```python
y_pred_test = model.predict(X_test)
```

```python
plt.imshow(X_test[1000])
```

```python
np.argmax(y_pred_test[1000])
```

```python
model.save('digit_brain2.h5')
```

```python
from PIL import Image
from keras.models import load_model

digit_brain = load_model('digit_brain2.h5')

img_path = r"E:\jupyter\DL\Chapter-7 (Image Classification Scripting)\Practice\Screenshot_8.png"
image = Image.open(img_path)
image = image.convert(mode='L')
image = image.resize((28,28))
image = np.array(image)
image
```

```python
def digit_pred(image):
    image = image/255.0
    image = np.expand_dims(image, axis=0)
    pred = digit_brain.predict(image)
    return np.argmax(pred)
```
