### 1. Import Necessary Libraries

```python
import keras
import pandas as pd
import matplotlib.pyplot as plt
```

### 2. Data Collection

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()
```

### 3. Data Understanding

```python
X_train.shape, y_train.shape, X_test.shape, y_test.shape
```

```python
assert X_train.shape == (60000, 28, 28)
```

```python
X_train[10000]
```

```python
pd.set_option("display.max_columns", 30)
pd.DataFrame(X_train[10000])
```

```python
plt.imshow(X_train[10000], cmap='grey')
```

### 4. Data Preparation

```python
# Normalize the data to faster processing
X_train.min(), X_train.max()
```

```python
X_train = X_train/255.0
X_test = X_test/255.0

X_train.min(), X_train.max()
```

```python
X_test.min(), X_test.max()
```

### 5. Model Building

```python
from keras.models import Sequential
from keras.layers import Flatten, Dense
```

```python
model = Sequential()
model.add(layer=Flatten())
model.add(Dense(units=100, activation='relu'))
model.add(Dense(units=10, activation='softmax'))

model.compile(optimizer='rmsprop', loss='sparse_categorical_crossentropy')
```

```python
model.summary()
```

### 6. Model Training | 7. Testing | 8. Evaluation

```python
model.fit(x=X_train, y=y_train, batch_size=32, epochs=10, validation_data=(X_test, y_test))
```

```python
model.summary()
```

### 9. Manual Prediction

```python
y_pred_test = model.predict(X_test)

plt.imshow(X_test[500])
```

```python
import numpy as np
np.argmax(y_pred_test[500])
```

### 10. Test outside new Data

```python
model.save('digit_brain.h5')
```

```python
from PIL import Image
from keras.models import load_model

digit_brain = load_model('digit_brain.h5')
```

```python
img_path = r"E:\jupyter\DL\Chapter-7 (Image Classification Scripting)\Practice\Screenshot_8.png"
image = Image.open(img_path)
image = image.convert(mode='L')
image = image.resize((28,28))
image = np.array(image)
image
```

```python
def digit_prediction(image):
    image = image/255.0
    image = np.expand_dims(a=image, axis=0)
    print(image.shape)
    predicted_img = digit_brain.predict(image)
    return np.argmax(predicted_img)
```

```python
digit_prediction(image)
```
