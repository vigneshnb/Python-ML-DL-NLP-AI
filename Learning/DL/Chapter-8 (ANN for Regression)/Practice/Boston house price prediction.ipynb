```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

from keras.models import Sequential, load_model
from keras.layers import Dense
from keras.datasets import boston_housing
```

```python
(x_train, y_train), (x_test, y_test) = boston_housing.load_data()
```

```python
x_train.shape, y_train.shape, x_test.shape, y_test.shape
```

```python
std_scalar = StandardScaler()

x_train = std_scalar.fit_transform(x_train)
x_test = std_scalar.fit_transform(x_test)
```

```python
model = Sequential()
model.add(Dense(64, activation='relu', input_shape=(13,)))
model.add(Dense(32, activation='relu'))
model.add(Dense(16, activation='relu'))
model.add(Dense(1, ))

model.compile(optimizer='adam', loss='mean_squared_error', metrics=['mean_absolute_error'])
```

```python
model.fit(x_train, y_train, batch_size=10, epochs=50, validation_split=0.2, verbose=2)
```

```python
new_data = np.array([[0.02, 18, 2.5, 0, 0.45, 6.5, 65, 4.5, 1, 300, 15, 390, 5]])

new_data = std_scalar.transform(new_data)

prediction = model.predict(new_data)

prediction
```
