```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
clg_data = pd.read_csv('college_list.csv')
clg_data

X = clg_data.drop('College_Name', axis=1)
X
```

```python
std_scalar = StandardScaler()
Scaled_X = std_scalar.fit_transform(X)
Scaled_X = pd.DataFrame(Scaled_X)
Scaled_X
```

```python
cl_model = KMeans(n_clusters=3, random_state=12)
cl_model.fit(Scaled_X)
cl_model.labels_
```

```python
cl_model.inertia_
```

```python
wcss = []
for i in range(1, 11):
    cl_model = KMeans(n_clusters=i, random_state=12)
    cl_model.fit(Scaled_X)
    wcss.append(cl_model.inertia_)
    
```

```python
plt.plot(range(1,11),wcss)
# plt.title('ELBOW METHOD')
# plt.xlabel('No of Clusters')
# plt.ylabel('WCSS')
plt.show()
```
