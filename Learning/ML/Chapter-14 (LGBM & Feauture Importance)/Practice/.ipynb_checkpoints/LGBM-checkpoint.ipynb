```python
import pandas as pd
import seaborn as sns

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import AdaBoostClassifier, GradientBoostingClassifier

from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix

import warnings
warnings.filterwarnings('ignore')
```

#### Import Dataset

```python
cc_df = pd.read_csv('credit_card_clean.csv')
cc_df
```

#### Data Understanding

```python
cc_df.shape
```

```python
cc_df.isna().sum()
```

```python
cc_df.dtypes
```

#### Data Preparation

```python
lbl_encoder = LabelEncoder()
cc_df['SEX'] = lbl_encoder.fit_transform(cc_df['SEX'])
cc_df['EDUCATION'] = lbl_encoder.fit_transform(cc_df['EDUCATION'])
cc_df['MARRIAGE'] = lbl_encoder.fit_transform(cc_df['MARRIAGE'])

# del cc_df['ID']
cc_df

X = cc_df.drop('DEFAULT', axis=1)
y = cc_df.DEFAULT

X.shape, y.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=True, random_state=123, stratify=y)
```

#### Model Building

```python
adb_classifier = AdaBoostClassifier()
gb_classifier = GradientBoostingClassifier(n_estimators=50)
xgb_classifier = XGBClassifier(n_estimators=200)
lgb_classifier = LGBMClassifier(n_estimators=200)
```

#### Model Training

```python
%%time
adb_classifier.fit(X_train, y_train)
```

```python
%%time
gb_classifier.fit(X_train, y_train)
```

```python
%%time
xgb_classifier.fit(X_train, y_train)
```

```python
%%time
lgb_classifier.fit(X_train, y_train)
```

#### Model Testing

```python
y_pred_agb = adb_classifier.predict(X_test)
y_pred_gb = gb_classifier.predict(X_test)
y_pred_xgb = xgb_classifier.predict(X_test)
y_pred_lgb = lgb_classifier.predict(X_test)
```

#### Model Evaluation

#### AdaBoost

```python
print('accuracy score: :', round(accuracy_score(y_test, y_pred_agb), 4))
print('precison score: :', round(precision_score(y_test, y_pred_agb), 4))
print('(recall score: :', round(recall_score(y_test, y_pred_agb), 4))
print('confusion matrix \n', confusion_matrix(y_test, y_pred_agb))
```

#### Gradient Boost

```python
print('accuracy score: :', round(accuracy_score(y_test, y_pred_gb), 4))
print('precison score: :', round(precision_score(y_test, y_pred_gb), 4))
print('(recall score: :', round(recall_score(y_test, y_pred_gb), 4))
print('confusion matrix \n', confusion_matrix(y_test, y_pred_gb))
```

#### Xtreme GradientBoost

```python
print('accuracy score: :', round(accuracy_score(y_test, y_pred_xgb), 4))
print('precison score: :', round(precision_score(y_test, y_pred_xgb), 4))
print('(recall score: :', round(recall_score(y_test, y_pred_xgb), 4))
print('confusion matrix \n', confusion_matrix(y_test, y_pred_xgb))
```

#### Light GBM

```python
print('accuracy score: :', round(accuracy_score(y_test, y_pred_lgb), 4))
print('precison score: :', round(precision_score(y_test, y_pred_lgb), 4))
print('(recall score: :', round(recall_score(y_test, y_pred_lgb), 4))
print('confusion matrix \n', confusion_matrix(y_test, y_pred_lgb))
```

#### Feature Importance

```python
adb_classifier_f = adb_classifier.feature_importances_
gb_classifier_f = gb_classifier.feature_importances_
xgb_classifier_f = xgb_classifier.feature_importances_
lgb_classifier_f = lgb_classifier.feature_importances_
```

```python
fi_list_df = pd.DataFrame(data = {
    'features': X.columns,
    'adb_fi': adb_classifier_f,
    'gb_fi': gb_classifier_f,
    'xgb_fi': xgb_classifier_f,
    'lgb_fi': lgb_classifier_f
})

fi_list_df

fi_list_df = fi_list_df.sort_values(by='gb_fi', ascending=False)
top_10 = fi_list_df.head(10)

top_10
```

```python

sns.barplot(data=top_10, x='gb_fi', y='features',)
```
